```python
from sklearn.model_selection import GroupShuffleSplit
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import (
RandomForestRegressor,
HistGradientBoostingRegressor,
RandomForestClassifier
)
from sklearn.metrics import (
mean_absolute_error,
mean_squared_error,
r2_score,
accuracy_score,
precision_score,
recall_score,
roc_auc_score,
confusion_matrix,
roc_curve
)
from IPython.display import display
sns.set_style("whitegrid")
print("All libraries imported successfully!")
```

```python
TARGET = "winPlacePerc"
TRAIN_SAMPLE_SIZE = 100000
model_df = df.sample(
n=min(TRAIN_SAMPLE_SIZE, len(df)),
random_state=42
).copy()
print("Model Dataset Shape:", model_df.shape)
display(model_df.head())
```

```python
y = model_df[TARGET].copy()
drop_columns = ["Id", "groupId", "matchId", "winPlacePerc"]
X = model_df.drop(
columns=drop_columns,
errors="ignore"
)
print("Independent Variables Shape:", X.shape)
print("Dependent Variable Shape:", y.shape)
```

```python
groups = model_df["matchId"]
gss = GroupShuffleSplit(
n_splits=1,
test_size=0.2,
random_state=42
)
train_index, test_index = next(
gss.split(X, y, groups=groups)
)
X_train = X.iloc[train_index].copy()
X_test = X.iloc[test_index].copy()
y_train = y.iloc[train_index].copy()
y_test = y.iloc[test_index].copy()
print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)
print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
print(
"Overlapping Matches:",
len(set(groups.iloc[train_index]) & set(groups.iloc[test_index]))
)
```

```python
categorical_columns = X_train.select_dtypes(
include=["object", "category"]
).columns.tolist()
print("Categorical Columns:", categorical_columns)
X_train = pd.get_dummies(
X_train,
columns=categorical_columns,
dtype=np.uint8
)
X_test = pd.get_dummies(
X_test,
columns=categorical_columns,
dtype=np.uint8
)
X_train, X_test = X_train.align(
X_test,
join="left",
axis=1,
fill_value=0
)
print("Categorical Encoding Completed!")
print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)
```

```python
X_train = X_train.replace([np.inf, -np.inf], np.nan)
X_test = X_test.replace([np.inf, -np.inf], np.nan)
medians = X_train.median(numeric_only=True).fillna(0)
X_train = X_train.fillna(medians).fillna(0)
X_test = X_test.fillna(medians).fillna(0)
print("Missing Values in Training Data:", X_train.isnull().sum().sum())
print("Missing Values in Testing Data:", X_test.isnull().sum().sum())
print("Preprocessing Completed Successfully!")
```

```python
dt_model = DecisionTreeRegressor(
max_depth=16,
min_samples_leaf=5,
random_state=42
)
dt_model.fit(X_train, y_train)
dt_predictions = dt_model.predict(X_test)
print("Decision Tree Model Training Completed!")
```

```python
rf_model = RandomForestRegressor(
n_estimators=100,
max_depth=15,
min_samples_leaf=2,
random_state=42,
n_jobs=-1
)
rf_model.fit(X_train, y_train)
rf_predictions = rf_model.predict(X_test)
print("Random Forest Model Training Completed!")
```

```python
gb_model = HistGradientBoostingRegressor(
max_iter=100,
learning_rate=0.1,
max_leaf_nodes=31,
random_state=42
)
gb_model.fit(X_train, y_train)
gb_predictions = gb_model.predict(X_test)
print("Gradient Boosting Model Training Completed!")
```

```python
def evaluate_model(model_name, actual, predicted):
    mae = mean_absolute_error(actual, predicted)
    mse = mean_squared_error(actual, predicted)
    rmse = np.sqrt(mse)
    r2 = r2_score(actual, predicted)

    return {
        "Model": model_name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    }
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

lr_model = LinearRegression()
lr_model.fit(X_train, y_train)
y_pred_lr = lr_model.predict(X_test)

dt_model = DecisionTreeRegressor(
    max_depth=15,
    random_state=42
)
dt_model.fit(X_train, y_train)
y_pred_dt = dt_model.predict(X_test)

rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)
rf_model.fit(X_train, y_train)
y_pred_rf = rf_model.predict(X_test)

print("All models trained successfully!")
```

```python
results = [
    evaluate_model("Linear Regression", y_test, y_pred_lr),
    evaluate_model("Decision Tree", y_test, y_pred_dt),
    evaluate_model("Random Forest", y_test, y_pred_rf)
]

results_df = pd.DataFrame(results)

print(results_df)
```

```python
y_win_train = np.isclose(
y_train,
1.0
).astype(int)
y_win_test = np.isclose(
y_test,
1.0
).astype(int)
print("First Place Class Distribution:")
print(pd.Series(y_win_train).value_counts())
print("\nTraining Target Classes:", np.unique(y_win_train))
print("Testing Target Classes:", np.unique(y_win_test))
```

```python
if len(np.unique(y_win_train)) == 2:
    win_classifier = RandomForestClassifier(
        n_estimators=80,
        max_depth=15,
        min_samples_leaf=3,
        class_weight="balanced_subsample",
        random_state=42,
        n_jobs=-1
    )
    win_classifier.fit(X_train, y_win_train)
    print("First-Place Classification Model Trained!")
else:
    win_classifier = None
    print(
        "Classification skipped: "
        "Training data contains only one class."
    )
```

```python
win_probabilities = None
if win_classifier is not None:
    if 1 in win_classifier.classes_:
        positive_index = list(
            win_classifier.classes_
        ).index(1)
        win_probabilities = (
            win_classifier.predict_proba(X_test)[:, positive_index]
        )
        print("First-Place Probability Predictions:")
        print(win_probabilities[:20])
    else:
        print("First-place class was not found.")
else:
    print("Win classifier is not available.")
```

```python
if win_classifier is not None:
    win_predictions = win_classifier.predict(X_test)
    print("========== FIRST-PLACE CLASSIFICATION ==========")
    print("Accuracy:", accuracy_score(y_win_test, win_predictions))
    print("Precision:", precision_score(
        y_win_test, win_predictions, zero_division=0
    ))
    print("Recall:", recall_score(
        y_win_test, win_predictions, zero_division=0
    ))
    if len(np.unique(y_win_test)) == 2 and win_probabilities is not None:
        auc = roc_auc_score(y_win_test, win_probabilities)
        print("ROC-AUC Score:", auc)
        fpr, tpr, thresholds = roc_curve(
            y_win_test, win_probabilities
        )
        plt.figure(figsize=(8, 6))
        plt.plot(fpr, tpr, label=f"ROC-AUC = {auc:.3f}")
        plt.plot([0, 1], [0, 1], linestyle="--")
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.title("ROC Curve - PUBG First-Place Prediction")
        plt.legend()
        plt.show()
    else:
        print(
            "ROC-AUC cannot be calculated because "
            "the test data contains only one class "
            "or probabilities are unavailable."
        )
    cm = confusion_matrix(
        y_win_test,
        win_predictions,
        labels=[0, 1]
    )
    plt.figure(figsize=(6, 5))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues"
    )
    plt.title("First-Place Classification Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()
else:
    print("Win classifier is not available.")
```

# REPORT ON CHALLENGES FACED

### 1. Introduction

The objective of this project is to analyze PUBG gameplay data and develop a Machine Learning model that predicts a player's final match placement (`winPlacePerc`) based on different gameplay statistics.

During the project, several challenges were encountered while performing data cleaning, exploratory data analysis, feature engineering, model building, and evaluation.

This report explains the major challenges, techniques used to overcome them, and the reasons for selecting those techniques.

---

### 2. Challenges Faced and Solutions

#### Challenge 1: Handling Missing Values

**Problem:**

The dataset contained missing values in some columns. Missing values can affect model performance and may cause errors during model training.

**Technique Used:**

- Identified missing values using `isnull().sum()`.
- Analyzed the percentage of missing values.
- Used median imputation for numerical columns where appropriate.
- Removed records where the target variable was missing.

**Reason:**

Median imputation was selected because it is less sensitive to extreme values compared to mean imputation. This helps preserve the overall distribution of numerical features.

#### Challenge 2: Removing Duplicate Records

**Problem:**

Duplicate records can introduce unnecessary repetition into the dataset and potentially affect model evaluation.

**Technique Used:**

Used the Pandas `drop_duplicates()` method to identify and remove duplicate rows.

**Reason:**

Removing duplicate records helps maintain data quality and prevents repeated observations from influencing the model unnecessarily.

#### Challenge 3: Handling Outliers

**Problem:**

Some numerical features, such as kills, damage dealt, distance travelled, and other gameplay statistics, may contain extreme values.

**Technique Used:**

- Used boxplots to identify potential outliers.
- Applied the Interquartile Range (IQR) method to investigate extreme observations.
- Retained legitimate extreme gameplay values instead of automatically deleting them.

**Reason:**

In PUBG, unusually high kills or damage values can represent genuine gameplay performance. Removing all extreme values could result in the loss of important information.

#### Challenge 4: Handling Categorical Variables

**Problem:**

The `matchType` column contains categorical values. Machine Learning algorithms generally require numerical input.

**Technique Used:**

Applied One-Hot Encoding using Pandas `get_dummies()`.

**Reason:**

One-Hot Encoding converts categorical values into numerical columns without introducing an artificial numerical relationship between categories.

#### Challenge 5: Removing Unnecessary Features

**Problem:**

The dataset contains identification columns such as `Id`, `groupId`, and `matchId`, which do not directly represent gameplay performance.

**Technique Used:**

Removed unnecessary identification columns before model training.

**Reason:**

These columns are identifiers rather than meaningful gameplay measurements. Removing them helps reduce irrelevant information and the possibility of the model learning meaningless patterns.

#### Challenge 6: Feature Selection

**Problem:**

The dataset contains multiple gameplay features. Not every feature contributes equally to predicting the final placement.

**Technique Used:**

- Analyzed feature correlations.
- Used correlation heatmaps.
- Examined Random Forest feature importance.

**Reason:**

Feature selection helps identify important gameplay factors, reduces unnecessary complexity, and improves model interpretability.

#### Challenge 7: Data Scaling

**Problem:**

Numerical features have different ranges. For example, kills, distance travelled, and damage dealt may have significantly different numerical values.

**Technique Used:**

Considered feature scaling for algorithms that are sensitive to differences in feature magnitudes.

Standardization can be applied using `StandardScaler` when required.

**Reason:**

Scaling helps algorithms such as Linear Regression with regularization and distance-based models work with features on comparable scales. Tree-based models such as Random Forest generally do not require scaling.

#### Challenge 8: Model Overfitting

**Problem:**

A Machine Learning model may perform well on training data but poorly on unseen data. This is known as overfitting.

**Technique Used:**

- Divided the dataset into training and testing sets using an 80:20 ratio.
- Compared training and testing performance.
- Considered model complexity and cross-validation.

**Reason:**

Testing on unseen data helps evaluate whether the model can generalize to new PUBG matches instead of simply memorizing the training data.

#### Challenge 9: Selecting the Best Machine Learning Model

**Problem:**

Different Machine Learning algorithms may produce different prediction results. Selecting the most suitable model was an important challenge.

**Technique Used:**

Trained and compared multiple regression algorithms:

1. Linear Regression
2. Random Forest Regressor
3. XGBoost Regressor

Evaluated the models using:

- Mean Absolute Error (MAE)
- Mean Squared Error (MSE)
- Root Mean Squared Error (RMSE)
- R-squared (R²)

**Reason:**

Using multiple evaluation metrics provides a better understanding of prediction errors and model performance.

The model with lower prediction errors and better generalization performance is considered more suitable for production.

#### Challenge 10: Understanding Important Gameplay Factors

**Problem:**

Predicting the final placement alone does not explain which gameplay activities have the greatest influence on a player's performance.

**Technique Used:**

Used Random Forest feature importance to identify influential gameplay features.

**Reason:**

Feature importance helps understand which gameplay statistics contribute most to the model's predictions and provides useful insights into player performance.
